```python
Missing_Route = df['Route'].mode()[0]
df.fillna({"Route":Missing_Route},inplace=True)
```

```python
df.isnull().sum()
Missing_Price = df['Price'].mean()
Missing_Price

df.fillna({'Price':Missing_Price},inplace=True)
```

# Graph Based Analysis

```python
import matplotlib.pyplot as plt
import seaborn as sns


```

```python

AirCount = df.groupby('Airline')['Airline'].count().sort_values()
AirCount.index
```

```python
# Count Distribution of Airline


ax = sns.countplot(x = df['Airline'],order=AirCount.index)
ax.bar_label(ax.containers[0])
plt.xticks(rotation=45)
plt.show()
```

```python
# Count Distribution according to Source

Source_Count = df.groupby('Source')['Source'].count().sort_values()

ax = sns.countplot(x=df['Source'],order=Source_Count.index)
ax.bar_label(ax.containers[0])
plt.show()
```

```python
# Price Distributoin agaisnt Airline
Air_Price = df.groupby('Airline')['Price'].sum().sort_values()

ax = sns.barplot(x=df['Airline'],y=df['Price'],estimator='sum',order=Air_Price.index,errorbar=None)
ax.bar_label(ax.containers[0],rotation=80,fmt="%0.0f")
plt.xticks(rotation=90)
plt.show()
```

```python
# Price with Source

Source_Price = df.groupby('Source')['Price'].sum().sort_values()
ax = sns.barplot(x=df['Source'],y=df['Price'],estimator='sum',errorbar=None,order = Source_Price.index)
ax.bar_label(ax.containers[0],rotation=80,fmt="%0.0f")
plt.show()
```

```python
TtlDf = df.groupby('Total_Stops')['Price'].agg(['mean','count']).sort_values(by='mean')
TtlDf

ax = sns.barplot(TtlDf['mean'])
ax.bar_label(ax.containers[0])

ax.twinx()
ax2 = sns.lineplot(TtlDf['count'],color='orange',marker="o")
plt.show()
```
