```python
df = pd.read_csv("E:/1.Eathans/Ethans/Data/Airline/Airline1.csv")
```

```python
Missing_Source = df['Source'].mode()[0]
print(Missing_Source)
df.fillna({'Source':Missing_Source},inplace=True)
```

```python
Missing_Destination = df['Destination'].mode()[0]
df.fillna({'Destination':Missing_Destination},inplace=True)
```

```python
Missing_Route = df['Route'].mode()[0]
df.fillna({"Route":Missing_Route},inplace=True)
```

```python
df.isnull().sum()
Missing_Price = df['Price'].mean()
Missing_Price

df.fillna({'Price':Missing_Price},inplace=True)
```
